```python
import pandas as pd

marks = pd.Series([80, 90, 70, 85, 95])

print("Sum:", marks.sum())
print("Mean:", marks.mean())
print("Maximum:", marks.max())
print("Minimum:", marks.min())
print("Count:", marks.count())
```

```python
import pandas as pd

marks = pd.Series([80, 90, 70, 85, 95])

print(marks.describe())
```

```python
import pandas as pd

marks = pd.Series(
    [85, 90, 65, 78, 95],
    index=["Sujitha", "Harika", "Ravi", "Priya", "Kiran"]
)

print("Student Marks:")
print(marks)

print("\nMarks above 80:")
print(marks[marks > 80])

print("\nAverage Marks:", marks.mean())
print("Highest Marks:", marks.max())
print("Lowest Marks:", marks.min())
```
